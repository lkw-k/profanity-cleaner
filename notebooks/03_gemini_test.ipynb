```python
#api 키 로드 테스트
import os
from dotenv import load_dotenv

# .env 파일 읽기 (노트북은 한 단계 위 폴더로 가야 .env에 닿음)
load_dotenv(dotenv_path='../.env')

api_key = os.getenv('GEMINI_API_KEY')
if api_key:
    print("API 키가 성공적으로 로드되었습니다.")
else:
    print("API 키를 찾을 수 없습니다. .env 파일을 확인하세요.")
```

```python
# 제미나이 호출 테스트
import google.generativeai as genai
#api 키 설정
genai.configure(api_key=api_key)
#모델 선택
model = genai.GenerativeModelmodel = genai.GenerativeModel('gemini-2.0-flash')
# 간단한 호출 테스트
response = model.generate_content("안녕 한국어로 짧게 인사해줘.")
print("제미나이 응답:", response.text)
```
