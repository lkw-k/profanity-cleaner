```python
import os
from dotenv import load_dotenv

load_dotenv(dotenv_path='../.env')

api_key = os.getenv('GROK_API_KEY')
if api_key and api_key != '여기에_그록_키_입력':
    print("Grok API 키 로드 성공")
else:
    print("API 키를 찾을 수 없습니다. .env 파일에 GROK_API_KEY를 입력하세요.")
```

```python
from openai import OpenAI

client = OpenAI(
    api_key=api_key,
    base_url="https://api.x.ai/v1"
)

response = client.chat.completions.create(
    model="grok-3-mini",
    messages=[{"role": "user", "content": "안녕 한국어로 짧게 인사해줘."}]
)
print("연결 성공:", response.choices[0].message.content)
```

```python
import google.generativeai as genai

genai.configure(api_key=api_key)

# generateContent를 지원하는 모델만 출력
print("사용 가능한 모델:")
for m in genai.list_models():
    if 'generateContent' in m.supported_generation_methods:
        print(f"  - {m.name}")
```
